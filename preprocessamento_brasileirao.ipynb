# Engenharia de features 


```python
df_raw['Date_parsed'] =pd.to_datetime(df_raw['Date'], format= '%d/%m/%Y')
df_raw = df_raw.sort_values('Date_parsed').reset_index(drop=True)
```

```python
# covenverter odds em Probabilidade 
df_raw['p_home'] = 1/df_raw['PSCH']
df_raw['p_draw'] = 1/df_raw['PSCD']
df_raw['p_away'] = 1/df_raw['PSCA']

```

```python
# remover o juice da casa de aposta
total = ( 
    df_raw['p_home'] +
    df_raw['p_draw'] +
    df_raw['p_away']
)
```

```python
df_raw["p_home"] /= total
df_raw["p_away"] /= total
df_raw["p_draw"] /= total
```

```python
# diferença porbabilidade casa e fora 
df_raw['dif_prob'] = df_raw['p_home'] - df_raw['p_away']
```

```python
df_raw['dif_draw'] = df_raw['p_home'] - df_raw['p_draw']
```

```python
df_raw['goal_diff'] = df_raw['HG'] - df_raw['AG']
```

```python
df_raw['goal_total'] = df_raw['HG'] + df_raw['AG']
```

# Categorizar as probabilidades do times 

```python
labels = [
    "Grande Zebra",
    "Zebra",
    "Leve Zebra",
    "Equilibrado",
    "Leve Favorito",
    "Favorito",
    "Super Favorito"
]
```

```python
df_raw["bucket_favoritismo"] = pd.qcut(
    df_raw["dif_prob"],
    q=7,
    labels=labels,
    duplicates="drop"
)
```

```python
df_raw.tail()
```

# Média de gols dos ultimos 5 jogos 

```python
# média de gols nos ultimos 5 jogos 

df_raw['home_last5_goals'] =(
    df_raw.groupby('Home')['HG']
    .transform(lambda s: s.shift().rolling(window=5, min_periods=1).mean())
    
)

df_raw['away_last5_goals'] =(
    df_raw.groupby('Away')['AG']
    .transform(lambda s: s.shift().rolling(window=5, min_periods=1).mean())
)
```

```python
# força da defesa 
df_raw['home_last5_concede'] = df_raw.groupby('Home')['AG'].transform(
    lambda s: s.shift().rolling(5, min_periods=1).mean()
)

df_raw['away_last5_concede'] = df_raw.groupby('Away')['HG'].transform(
    lambda s: s.shift().rolling(5, min_periods=1).mean()
)
```

# Média exponencial do gols EWMA

```python
df_raw['home_ewma_5_goals'] = (
    df_raw.groupby('Home')['HG']
    .transform(lambda s : s.shift().ewm(span = 10, adjust = False).mean())
)

df_raw['away_ewma_5_goals'] = (
    df_raw.groupby('Away')['AG']
    .transform(lambda s : s.shift().ewm(span = 10, adjust = False).mean())
)
```

# Winrate dos ultimos 5 jogos 

```python
res_map ={ 'H' :2, 'D': 1, 'A' : 0}
df_raw['Res_num'] = df_raw['Res'].map(res_map)
```

```python
df_raw['home_last5_winrate'] =(
    df_raw.groupby('Home')['Res_num']
    .transform(lambda s: s.shift().rolling(5, min_periods= 1).apply(lambda x : (x==2).mean()))
)

df_raw['away_last5_winrate'] =(
    df_raw.groupby('Away')['Res_num']
    .transform(lambda s: s.shift().rolling(5, min_periods= 1).apply(lambda x : (x==0).mean()))
)
```

```python
df_raw["home_points"] = np.select(
    [
        df_raw["Res_num"] == 2,
        df_raw["Res_num"] == 1,
        df_raw["Res_num"] == 0
    ],
    [
        3,
        1,
        0
    ]
)

df_raw["away_points"] = np.select(
    [
        df_raw["Res_num"] == 0,
        df_raw["Res_num"] == 1,
        df_raw["Res_num"] == 2
    ],
    [
        3,
        1,
        0
    ]
)
```

# PONTOS NAS ULTIMAS 5 PARTIDAS PPG

```python
df_raw["home_last5_ppg"] = (
    df_raw
    .groupby("Home")["home_points"]
    .transform(
        lambda s: s.shift().rolling(5, min_periods=1).mean()
    )
)


df_raw["away_last5_ppg"] = (
    df_raw
    .groupby("Away")["away_points"]
    .transform(
        lambda s: s.shift().rolling(5, min_periods=1).mean()
    )
)
```

# CALCULANDO O CAOS "ENTROPIA"

```python
H_MAX = np.log2(3)

def calc_entropy(row, normalize=True):

    p = np.array([
        row["p_home"],
        row["p_draw"],
        row["p_away"]
    ])

    p = p[p > 0]

    h = -(p * np.log2(p)).sum()

    if normalize:
        h /= H_MAX

    return h

df_raw["entropy"] = df_raw.apply(calc_entropy, axis=1)
```

| Entropia | Interpretação                   |
| -------: | ------------------------------- |
|     0.00 | Resultado praticamente definido |
|     0.20 | Super favorito                  |
|     0.40 | Favorito                        |
|     0.60 | Leve favorito                   |
|     0.80 | Jogo equilibrado                |
|     1.00 | Máxima incerteza                |
